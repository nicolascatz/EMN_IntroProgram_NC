```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0, 10, 0.1)


f1=np.sin(x)
f2=np.sin(5*x)

plt.plot(f1,'b')
plt.plot(f2,'r')

plt.show()

plt.plot(f1+f2)


```

design a filter with scipy 
Notch Filter for a very narrow freq band

```python
from scipy.signal import iirnotch, filtfilt
signal=f1+f2
b,a = iirnotch(1.59, 0.3, 50)
signal_filtered = filtfilt(b, a, signal)

plt.plot(signal_filtered,'r')
plt.plot(signal,'b')
plt.show()



```

```python

from scipy.signal import butter, filtfilt


b,a=butter(4, [0.1, 6], fs=50, btype='band')
signal_filtered = filtfilt(b, a, signal)

plt.plot(signal_filtered,'r')
plt.plot(signal,'b')
```
