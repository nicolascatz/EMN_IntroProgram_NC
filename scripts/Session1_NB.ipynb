```python
workingtime=[10,2,9,30,40,6,15]
grade=[12,6,7,16,18,6,12]
studentnames=["Nicolas","John","Mary","Peter","Paul","Anna","Tom"]

data = np.array([workingtime,grade]) # creating a 2D array with two rows and seven columns


data
```

```python



print(data)
print(data.shape) #shape of the array

data_transposed=np.transpose(data) #transpose of the array
# data_transposed = data.T # transposing the array to have seven rows and two columns   

```

```python
print(data_transposed)

print("mean of each column:")
print(data.mean(axis=1)) # mean value of each column
print("max of each column:")
print(data.max(axis=1)) # maximum value of each column
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5.5))
scatter = ax.scatter(
    data_transposed[:, 0],
    data_transposed[:, 1],
    c=data_transposed[:, 1],
    cmap="viridis",
    s=90,
    alpha=0.9,
    edgecolors="white",
    linewidths=1.2,
)

for name, working_hours, student_grade in zip(
    studentnames, data_transposed[:, 0], data_transposed[:, 1]
):
    ax.annotate(
        name,
        (working_hours, student_grade),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=9,
    )

ax.set_xlabel("Working time (hours)", labelpad=8)
ax.set_ylabel("Grade", labelpad=8)
ax.set_title("Working Time and Grade", pad=14, weight="bold")
ax.set_xlim(0, max(data_transposed[:, 0]) + 5)
ax.set_ylim(0, max(data_transposed[:, 1]) + 3)
ax.set_xticks(range(0, int(max(data_transposed[:, 0])) + 6, 5))
ax.grid(True, linestyle="--", alpha=0.3)
ax.spines[["top", "right"]].set_visible(False)
fig.colorbar(scatter, ax=ax, label="Grade")
fig.tight_layout()
plt.show()
```

```python
import scipy.stats as stats

    
sns.boxenplot(data=data,x='Dx',y='Attention') #boxenplot for the Speed variable
AnovaResults=stats.f_oneway(data[data['Dx']=='Control']['Attention'],data[data['Dx']=='Schizophrenia']['Attention'],data[data['Dx']=='Schizoaffective']['Attention'])
ttest0=stats.ttest_ind(data[data['Dx']=='Control']['Attention'],data[data['Dx']=='Schizoaffective']['Attention'])
ttest1=stats.ttest_ind(data[data['Dx']=='Control']['Attention'],data[data['Dx']=='Schizophrenia']['Attention'])
ttest2=stats.ttest_ind(data[data['Dx']=='Schizophrenia']['Attention'],data[data['Dx']=='Schizoaffective']['Attention'])

print("ANOVA results:", AnovaResults)
print("T-test schizophrenia vs schizoaffective:", ttest2)
print("T-test control vs schizoaffective:", ttest0)
print("T-test control vs schizophrenia:", ttest1)



```
