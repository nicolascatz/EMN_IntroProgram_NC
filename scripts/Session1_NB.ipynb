```python
workingtime=[10,2,9,30,40,6,15]
grade=[12,6,7,16,18,6,12]
studentnames=["Nicolas","John","Mary","Peter","Paul","Anna","Tom"]

data = np.array([workingtime,grade]) # creating a 2D array with two rows and seven columns
print(data)
print(data.shape) #shape of the array

data_transposed = data.T # transposing the array to have seven rows and two columns   
print(data_transposed)
print(data_transposed.shape) #shape of the array

import matplotlib.pyplot as plt # importing the matplotlib library for plotting 
    # if not existing then first install it with the command: pip install matplotlib
    
plt.plot(workingtime,grade,'.') # plotting the working time vs grade    
plt.xlabel('Working time (hours)') # labeling the x-axis
plt.ylabel('Grade') # labeling the y-axis
plt.title('Working time vs Grade') # adding a title to the plot
```
